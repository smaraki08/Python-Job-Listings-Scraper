```python
import requests
url = "https://realpython.github.io/fake-jobs/"
print(requests.get(url).text[:100])
```

```python
from bs4 import BeautifulSoup

# Parse the connection data into searchable HTML
soup = BeautifulSoup(requests.get(url).content, "html.parser")

# Find all job container elements on the webpage
job_elements = soup.find_all("div", class_="card-content")

# Print the total number of job listings found on the page to confirm
print(f"Successfully found {len(job_elements)} job listings!")
```

```python
# Loop through each of the 100 job card containers
for job in job_elements:
    # Extract the text inside the specific HTML sub-tags
    title = job.find("h2", class_="title").text.strip()
    company = job.find("h3", class_="company").text.strip()
    location = job.find("p", class_="location").text.strip()

    # Print the formatted results cleanly
    print(f"Job Title: {title}")
    print(f"Company:   {company}")
    print(f"Location:  {location}")
    print("-" * 40)
```
